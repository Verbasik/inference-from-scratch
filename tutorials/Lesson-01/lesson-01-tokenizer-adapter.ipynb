Токенизация — это преобразование текста (включая диалоговые сообщения) в последовательность идентификаторов словаря модели. Для совместимости с семейством Qwen3 и экосистемой Hugging Face мы создаём тонкий, но строгий адаптер, который:

- Единообразно кодирует строки, батчи строк и чат‑сообщения.
- Гарантирует корректные `pad_token_id` и `eos_token_id` (при отсутствии `pad` использует `eos`).
- Сохраняет тип тензоров как `torch.int32` (память/совместимость с рантаймом).
- Поддерживает `apply_chat_template` и обратимое `decode(encode(x))` на практических кейсах.

Где:
- $B$ — размер батча;
- $L$ — целевая длина последовательности после паддинга;
- Память для int32: 4 байта на токен.

<h2 id="section-bazovyj-adapter-tokenizatora" style="color: #ffffff; font-size: 1.8em; font-weight: 700; border-bottom: 3px solid #667eea; padding-bottom: 12px; margin-top: 40px; margin-bottom: 20px; letter-spacing: -0.01em;">
  <span style="background: linear-gradient(135deg, #667eea, #764ba2); -webkit-background-clip: text; -webkit-text-fill-color: transparent; margin-right: 8px;">📝</span> Базовый адаптер токенизатора
</h2>

<div style="background: #ebf8ff; border-left: 4px solid #63b3ed; padding: 16px 20px; border-radius: 0 10px 10px 0; margin: 12px 0; color: #2d3748;">
  <div style="font-weight: 700; color: #2b6cb0; margin-bottom: 6px; font-size: 0.95em;">📌 Определение</div>
  <div style="line-height: 1.7;">Адаптер — класс, выравнивающий интерфейс токенизатора: <code>encode</code>, <code>batch_encode</code>, <code>decode</code>, <code>apply_chat_template</code>, а также обеспечивающий корректные <code>pad/eos</code> и типы тензоров.</div>

<div style="background: #1a202c; color: #a0aec0; padding: 8px 16px; border-radius: 10px 10px 0 0; margin-bottom: -15px; font-family: 'Fira Code', monospace; font-size: 0.8em; display: flex; justify-content: space-between; align-items: center;">

    - nano-infer/python/infer/utils/hf.py;<br>

````python
class TokenizerAdapter:
    """
    Description:
    ---------------
        Унифицированный адаптер для токенизаторов Hugging Face
        (например, для семейства Qwen3). Обеспечивает единый API для
        кодирования строк, батчей и чат‑сообщений, а также обратимое
        декодирование и корректные типы/токены (pad/eos).

    Args:
    ---------------
        model_id: Идентификатор модели/токенизатора в HF Hub.
        use_fast: Использовать fast‑реализацию (при наличии).
        trust_remote_code: Доверять кастомному коду из репозитория.
        add_eos_token: Принудительно добавлять EOS при кодировании.

    Returns:
    ---------------
        Экземпляр адаптера с инициализированным токенизатором.

    Raises:
    ---------------
        ValueError: Если не удаётся корректно установить pad/eos токены.
        OSError: Если загрузка токенизатора с HF завершилась ошибкой.

    Examples:
    ---------------
        >>> adapter = TokenizerAdapter("Qwen/Qwen3-8B")
        >>> ids = adapter.encode("Привет")
        >>> text = adapter.decode(ids)
        >>> assert "Привет" in text
    """

    def __init__(
        self,
        model_id: str = DEFAULT_MODEL_ID,
        use_fast: bool = True,
        trust_remote_code: bool = True,
        add_eos_token: bool = False,
    ) -> None:
        
        # Сохраняем параметры и флаги
        self.model_id: str = model_id             # Идентификатор модели
        self.add_eos_token: bool = add_eos_token  # Флаг добавления EOS

        try:
            # Пытаемся загрузить токенизатор с HF
            self.tokenizer = AutoTokenizer.from_pretrained(
                model_id,
                use_fast=use_fast,
                trust_remote_code=trust_remote_code,
            )
        except OSError as e:
            # Преобразуем исключение в информативную ошибку
            raise OSError(
                f"Не удалось загрузить токенизатор: {model_id}. Ошибка: {e}"
            ) from e

        # Читаем EOS из токенизатора (если определён)
        eos_id: Optional[int] = getattr(self.tokenizer, "eos_token_id", None)
        # Читаем PAD из токенизатора (если определён)
        pad_id: Optional[int] = getattr(self.tokenizer, "pad_token_id", None)

        # Если нет pad — используем eos в качестве pad (распространённая практика)
        if pad_id is None:
            # Если eos не определён — это критично для совместимости
            if eos_id is None:
                raise ValueError(
                    "Отсутствуют и pad_token_id, и eos_token_id — задайте хотя бы EOS"
                )
            # Выставляем pad равным eos
            self.tokenizer.pad_token = self.tokenizer.eos_token  # type: ignore
            pad_id = self.tokenizer.pad_token_id                 # type: ignore

        # Сохраняем идентификаторы токенов как атрибуты экземпляра
        self.eos_token_id: Optional[int] = eos_id  # EOS может быть None
        self.pad_token_id: int = int(pad_id)       # PAD теперь точно определён

        print(f"[TokenizerAdapter] Итог: eos_token_id={self.eos_token_id}, pad_token_id={self.pad_token_id}")
        print(f"[TokenizerAdapter] Размер словаря: {self.tokenizer.vocab_size:,} токенов")

    def encode(
        self,
        text: str,
        *,
        add_special_tokens: bool = True,
        max_length: Optional[int] = None,
        padding: bool = False,
        truncation: bool = False,
        return_tensors: Literal["pt"] = "pt",
        dtype: torch.dtype = DEFAULT_DTYPE,
    ) -> torch.Tensor:
        """
        Description:
        ---------------
            Кодирует одиночную строку в тензор индексов токенов.

        Args:
        ---------------
            text: Исходная текстовая строка для кодирования.
            add_special_tokens: Добавлять спецтокены токенизатора.
            max_length: Целевая длина (для паддинга/усечения).
            padding: Применять паддинг до max_length.
            truncation: Применять усечение до max_length.
            return_tensors: Формат возврата (поддерживается 'pt').
            dtype: Тип данных тензора (по умолчанию int32).

        Returns:
        ---------------
            torch.Tensor: Тензор индексов токенов формы (L,).

        Raises:
        ---------------
            ValueError: Если запрошен неподдерживаемый return_tensors.

        Examples:
        ---------------
            >>> adapter = TokenizerAdapter()
            >>> ids = adapter.encode("Привет")
            >>> ids.dtype == DEFAULT_DTYPE
            True
        """
        # Проверяем поддерживаемый формат возврата
        if return_tensors != "pt":
            raise ValueError("Поддерживается только return_tensors='pt'")

        # Подготавливаем параметры паддинга
        pad_mode: Union[bool, str] = False      # По умолчанию без паддинга
        if padding and max_length is not None:
            pad_mode = "max_length"             # Паддинг до max_length

        # Вызываем токенизатор HF
        encoded: Dict[str, torch.Tensor] = self.tokenizer(
            text,
            add_special_tokens=add_special_tokens or self.add_eos_token,
            padding=pad_mode,
            truncation=truncation,
            max_length=max_length,
            return_tensors="pt",
        )

        # Достаём тензор индексов и приводим тип к целевому
        input_ids: torch.Tensor = encoded["input_ids"].to(dtype=dtype)
        input_ids = input_ids.squeeze(0)

        print(f"[encode] Результат:  {input_ids.shape} токенов, dtype={input_ids.dtype}")
        print(f"[encode] ID токенов: {input_ids.tolist()}")
        print(f"[encode] Токены:     {self.tokenizer.convert_ids_to_tokens(input_ids.tolist())}")
        
        # Удаляем размерность батча (1, L) -> (L,)
        return input_ids.squeeze(0)

    def batch_encode(
        self,
        texts: List[str],
        *,
        add_special_tokens: bool = True,
        max_length: Optional[int] = None,
        padding: bool = True,
        truncation: bool = True,
        return_tensors: Literal["pt"] = "pt",
        dtype: torch.dtype = DEFAULT_DTYPE,
    ) -> Dict[str, torch.Tensor]:
        """
        Description:
        ---------------
            Кодирует список строк в батч тензоров (input_ids,
            attention_mask) с паддингом.

        Args:
        ---------------
            texts: Список строк для кодирования.
            add_special_tokens: Добавлять спецтокены токенизатора.
            max_length: Целевая длина последовательности.
            padding: Включить паддинг (обычно True для батчей).
            truncation: Включить усечение по max_length.
            return_tensors: Формат возврата ('pt').
            dtype: Тип данных выходных тензоров.

        Returns:
        ---------------
            Dict[str, torch.Tensor]: Поля 'input_ids', 'attention_mask'.

        Raises:
        ---------------
            ValueError: Если список пуст или неподдерживаемый формат.

        Examples:
        ---------------
            >>> adapter = TokenizerAdapter()
            >>> batch = adapter.batch_encode(["a", "b"], max_length=8)
            >>> batch["input_ids"].dtype == DEFAULT_DTYPE
            True
        """
        # Базовые проверки входных данных
        if not texts:
            raise ValueError("Список текстов пуст — нечего кодировать")
        if return_tensors != "pt":
            raise ValueError("Поддерживается только return_tensors='pt'")

        # Режим паддинга: для батча по умолчанию включаем
        pad_mode: Union[bool, str] = padding
        if padding and max_length is not None:
            pad_mode = "max_length"  # Жёсткая длина для стабильных батчей

        # Кодируем батч строк
        encoded = self.tokenizer(
            texts,
            add_special_tokens=add_special_tokens or self.add_eos_token,
            padding=pad_mode,
            truncation=truncation,
            max_length=max_length,
            return_tensors="pt",
        )

        # Приводим типы к целевому и возвращаем словарь тензоров
        outputs: Dict[str, torch.Tensor] = {
            k: v.to(dtype=dtype) for k, v in encoded.items()
        }

        ids = outputs["input_ids"]
        mask = outputs.get("attention_mask")
        print(f"[batch_encode] input_ids shape: {ids.shape}, dtype={ids.dtype}")
        
        if mask is not None:
            pad_counts = (mask == 0).sum(dim=1).tolist()
            print(f"[batch_encode] PAD токенов на строку: {pad_counts}")

        return outputs

    def apply_chat_template(
        self,
        messages: List[Dict[str, str]],
        *,
        add_generation_prompt: bool = True,
    ) -> str:
        """
        Description:
        ---------------
            Формирует текстовую подсказку из списка чат‑сообщений,
            используя шаблон токенизатора (ChatML‑подобный формат).

        Args:
        ---------------
            messages: Список словарей вида {"role", "content"}.
            add_generation_prompt: Добавить заготовку для ответа модели.

        Returns:
        ---------------
            str: Отформатированная строка для последующего кодирования.

        Raises:
        ---------------
            ValueError: Если сообщения имеют неверный формат.

        Examples:
        ---------------
            >>> adapter = TokenizerAdapter()
            >>> prompt = adapter.apply_chat_template([
            ...     {"role": "user", "content": "Привет"}
            ... ])
            >>> isinstance(prompt, str)
            True
        """
        # Минимальная валидация формата сообщений
        if not isinstance(messages, list) or not all(
            isinstance(m, dict) and "role" in m and "content" in m
            for m in messages
        ):
            raise ValueError(
                "messages должны быть списком словарей с ключами 'role' и 'content'"
            )

        # Применяем chat template токенизатора: получаем строку
        formatted: str = self.tokenizer.apply_chat_template(
            messages,
            tokenize=False,
            add_generation_prompt=add_generation_prompt,
        )

        print(f"[apply_chat_template] Итоговый промпт ({len(formatted)} символов):")
        print(f"  {formatted!r}")

        return formatted

    def encode_chat(
        self,
        messages: List[Dict[str, str]],
        *,
        max_length: Optional[int] = None,
        padding: bool = False,
        truncation: bool = False,
        dtype: torch.dtype = DEFAULT_DTYPE,
    ) -> torch.Tensor:
        """
        Description:
        ---------------
            Кодирует чат‑сообщения: сначала применяет шаблон, затем
            кодирует итоговую строку в тензор индексов токенов.

        Args:
        ---------------
            messages: Список словарей {"role", "content"}.
            max_length: Целевая длина.
            padding: Включить паддинг.
            truncation: Включить усечение.
            dtype: Тип данных тензора результата.

        Returns:
        ---------------
            torch.Tensor: Последовательность индексов токенов (L,).

        Examples:
        ---------------
            >>> adapter = TokenizerAdapter()
            >>> ids = adapter.encode_chat([
            ...     {"role": "user", "content": "Как дела?"}
            ... ])
            >>> ids.ndim
            1
        """
        # Формируем подсказку по шаблону
        prompt: str = self.apply_chat_template(messages)
        # Кодируем итоговую строку тем же механизмом, что и одиночный текст
        ids = self.encode(
            prompt,
            max_length=max_length,
            padding=padding,
            truncation=truncation,
            dtype=dtype,
        )

        print(f"[encode_chat] Готово: {ids.shape[0]} токенов для {len(messages)} сообщений")
        return ids

    def decode(
        self,
        ids: Union[List[int], torch.Tensor],
        *,
        skip_special_tokens: bool = True,
    ) -> str:
        """
        Description:
        ---------------
            Декодирует последовательность индексов токенов обратно в
            строку, при необходимости пропуская спец‑токены.

        Args:
        ---------------
            ids: Список индексов или тензор токенов.
            skip_special_tokens: Пропускать ли спец‑токены при декоде.

        Returns:
        ---------------
            str: Восстановленная строка текста.

        Examples:
        ---------------
            >>> adapter = TokenizerAdapter()
            >>> s = adapter.decode(adapter.encode("Тест"))
            >>> "Тест" in s
            True
        """
        # Приводим вход к списку целых чисел
        if isinstance(ids, torch.Tensor):
            as_list: List[int] = [int(x) for x in ids.view(-1).tolist()]
        else:
            as_list = [int(x) for x in ids]

        # Декодируем через токенизатор HF
        result = self.tokenizer.decode(
            as_list, skip_special_tokens=skip_special_tokens
        )
        
        print(f"[decode] Результат: {result!r}")
        return result
````

````python
# Демонстрация форматирования диалога через chat template
def demo_build_prompt() -> str:
    """
    Description:
    ---------------
        Строит текст подсказки из сообщений, используя адаптер.

    Args:
    ---------------
        None

    Returns:
    ---------------
        Готовая строка подсказки для кодирования/генерации.

    Raises:
    ---------------
        Нет явных исключений в этом примере.

    Examples:
    ---------------
        >>> s = demo_build_prompt()
        >>> isinstance(s, str)
        True
    """
    # Создаём адаптер с моделью по умолчанию
    adapter = TokenizerAdapter(model_id=DEFAULT_MODEL_ID)
    # Формируем список сообщений в формате role/content
    messages = [
        {"role": "system", "content": "Вы — полезный ассистент."},
        {"role": "user", "content": "Объясни токенизацию простыми словами."},
    ]
    # Генерируем строку подсказки через chat template
    prompt = adapter.apply_chat_template(messages)
    # Возвращаем строку — её можно передать в encode или генератор
    return prompt
````

````python
# Демонстрация батч‑кодирования с паддингом до max_length
def demo_batch_encode(max_length: int = 36) -> Dict[str, torch.Tensor]:
    """
    Description:
    ---------------
        Кодирует список строк в батарее примеров и возвращает
        словарь тензоров с одинаковой длиной.

    Args:
    ---------------
        max_length: Целевая длина последовательности после паддинга.

    Returns:
    ---------------
        Dict[str, torch.Tensor]: Поля 'input_ids', 'attention_mask'.

    Raises:
    ---------------
        ValueError: Если внутренняя валидация входных данных не проходит.

    Examples:
    ---------------
        >>> out = demo_batch_encode(max_length=16)
        >>> out["input_ids"].shape[1]
        16
    """
    # Инициализируем адаптер токенизатора
    adapter = TokenizerAdapter(model_id=DEFAULT_MODEL_ID)
    # Готовим список входных строк
    examples: List[str] = [
        "Привет!",                                            # Короткая строка
        "Расскажите, пожалуйста, как работает токенизация?",  # Длиннее
        "OK",                                                 # Очень короткая
    ]
    # Вызываем batch_encode с паддингом до max_length
    batch = adapter.batch_encode(
        examples,
        max_length=max_length,
        padding=True,
        truncation=True,
    )
    # Возвращаем тензоры для дальнейшей обработки
    return batch
````

**Три строки — три разные судьбы**

`"OK"` — всего 1 токен. `"Привет!"` — 4 токена. `"Расскажите, пожалуйста, как работает токенизация?"` — 19 токенов. Токенизатор Qwen3 работает на субсловных единицах (BPE), поэтому русское слово может разбиться на 2–5 кусков: `"Расскажите"` → `["▁Рас", "ска", "жите"]`.

---

**Зачем нужен PAD токен**

Матрица тензора прямоугольная — все строки обязаны иметь одинаковую длину. Чтобы привести их к единой длине `max_length=36`, токенизатор добавляет специальный токен-заглушку `<pad>` с `id=151643`. У `"OK"` таких заглушек 35, у длинной строки — 17.

---

**`attention_mask` — это фильтр**

Здесь и нужна маска. Модель Transformer обрабатывает весь батч целиком, включая паддинг. Без маски она бы «смотрела» на заглушки так же, как на реальные слова, и усредняла бы смысл с мусором.

- `1` → этот токен настоящий, учитывай его при вычислении внимания
- `0` → это паддинг, не трать на него внимание

На практике модель перед softmax в слое attention добавляет к позициям с маской `0` очень большое отрицательное число (`-inf`), что делает их вес после softmax практически нулевым.

---

**Почему `pad_token_id = 151643`, а `vocab_size = 151 643`?**

Обратите внимание на вывод `__init__`: размер словаря равен 151 643, а `pad_token_id = 151 643` — это индекс за пределами словаря (он равен `vocab_size`, то есть `vocab_size + 0`, если считать с нуля). Это намеренный выбор Qwen3: выделить для PAD отдельный слот вне основного словаря, чтобы ни один реальный токен с ним не совпадал. В логах `__init__` видно, что `eos_token_id=151645` и `pad_token_id=151643` — оба выставлены явно в конфиге модели, поэтому фолбэк `pad = eos` не сработал.

````python
# Простой стрим‑декодер и проверка на round‑trip

_SAFE_CHARS: set[str] = {" ", ".", ",", "!", "?", ";", ":", "\n"}


def _is_boundary_token(adapter: TokenizerAdapter, token_id: int) -> bool:
    """
    Description:
    ---------------
        Проверяет, является ли одиночный токен безопасной точкой разрыва.
        Декодирует ровно один токен без skip_special_tokens, чтобы не
        потерять пробельные токены (они тоже спецсимволы у части моделей).
    Args:
    ---------------
        adapter   : Инициализированный TokenizerAdapter.
        token_id  : ID токена для проверки.
    Returns:
    ---------------
        True, если декодированный текст токена состоит только из символов
        из _SAFE_CHARS (например: ' ', '.', ',').
    """
    decoded = adapter.tokenizer.decode([token_id], skip_special_tokens=False)
    return bool(decoded) and all(ch in _SAFE_CHARS for ch in decoded)


def safe_stream_decode(
    adapter: TokenizerAdapter,
    buffer: List[int],
    *,
    force_every: int = 8,
) -> Tuple[str, List[int]]:
    """
    Description:
    ---------------
        Декодирует накопленные токены, если появился токен-разделитель
        (пробел/пунктуация) или накопилось force_every токенов.

        Разбиение происходит по индексу токена в буфере, а не по символьной
        позиции в строке — это исключает потерю токенов при BPE-декодировании.
    Args:
    ---------------
        adapter    : Экземпляр TokenizerAdapter для декодирования.
        buffer     : Список накопленных токенов.
        force_every: Принудительная частота декодирования по числу токенов.
    Returns:
    ---------------
        Кортеж (фрагмент текста для вывода, оставшийся буфер токенов).
        Если ещё рано декодировать — возвращает ("", buffer) без изменений.
    Raises:
    ---------------
        ValueError: Если передан неинициализированный адаптер или буфер.
    Examples:
    ---------------
        >>> adapter = TokenizerAdapter()
        >>> text, rest = safe_stream_decode(adapter, [1, 2, 3])
        >>> isinstance(text, str)
        True
    """
    if adapter is None:
        raise ValueError("Адаптер не должен быть None")
    if buffer is None:
        raise ValueError("Буфер токенов не должен быть None")
    if not buffer:
        return "", []

    print(f"[safe_stream_decode] буфер={buffer} (len={len(buffer)}, force_every={force_every})")

    # ── Ветка 1: принудительный сброс по размеру буфера ─────────────────────
    if len(buffer) >= force_every:
        text = adapter.decode(buffer)
        print(f"[safe_stream_decode] force flush → {text!r}")
        return text, []

    # ── Ветка 2: ищем последний токен-разделитель в буфере ──────────────────
    # Идём с конца, чтобы декодировать максимально длинный готовый фрагмент
    last_boundary: int = -1
    for i in range(len(buffer) - 1, -1, -1):
        if _is_boundary_token(adapter, buffer[i]):
            last_boundary = i
            break

    if last_boundary >= 0:
        to_decode = buffer[: last_boundary + 1]   # токены до разделителя включительно
        remaining = buffer[last_boundary + 1 :]   # токены после границы
        text = adapter.decode(to_decode)
        print(f"[safe_stream_decode] boundary at idx={last_boundary} → {text!r}, "
              f"остаток={remaining}")
        return text, remaining

    # ── Ветка 3: разделителя нет, буфер мал — ждём ──────────────────────────
    print("[safe_stream_decode] ждём ещё токенов...")
    return "", buffer


def demo_safe_stream_decode() -> None:
    """
    Description:
    ---------------
        Демонстрирует работу safe_stream_decode в трёх сценариях:
        1. force_every — принудительный сброс по порогу токенов.
        2. Разделитель найден — декодируем до него.
        3. Разделитель не найден — буфер остаётся нетронутым.
        4. Имитация реального стрима — токены приходят по одному.
    """
    adapter = TokenizerAdapter(model_id=DEFAULT_MODEL_ID)

    test_phrase = "Привет, мир. Как дела?"
    all_ids: List[int] = adapter.encode(test_phrase).tolist()

    print(f"\n{'='*60}")
    print(f"Тестовая фраза : {test_phrase!r}")
    print(f"Все токены     : {all_ids}")
    print(f"Всего токенов  : {len(all_ids)}")
    print(f"{'='*60}")

    # ── Сценарий 1: buffer >= force_every ───────────────────────────────────
    print("\n[ Сценарий 1 ] force_every=4, буфер = первые 4 токена")
    buffer_1: List[int] = all_ids[:4]
    print(f"  Буфер до вызова : {buffer_1}")
    text_1, rest_1 = safe_stream_decode(adapter, buffer_1, force_every=4)
    print(f"  Декодировано    : {text_1!r}")
    print(f"  Остаток буфера  : {rest_1}  (пуст — буфер сброшен)")

    # ── Сценарий 2: разделитель найден ──────────────────────────────────────
    print("\n[ Сценарий 2 ] force_every=20, буфер = первые 6 токенов")
    buffer_2: List[int] = all_ids[:6]
    print(f"  Буфер до вызова : {buffer_2}")
    print(f"  Промежуточный текст : {adapter.decode(buffer_2)!r}")
    text_2, rest_2 = safe_stream_decode(adapter, buffer_2, force_every=20)
    print(f"  Декодировано    : {text_2!r}")
    print(f"  Остаток буфера  : {rest_2}")

    # ── Сценарий 3: разделитель не найден, буфер копится ────────────────────
    print("\n[ Сценарий 3 ] force_every=20, буфер = первые 2 токена (нет разделителя)")
    buffer_3: List[int] = all_ids[:2]
    print(f"  Буфер до вызова : {buffer_3}")
    print(f"  Промежуточный текст : {adapter.decode(buffer_3)!r}")
    text_3, rest_3 = safe_stream_decode(adapter, buffer_3, force_every=20)
    print(f"  Декодировано    : {text_3!r}  (пусто — ждём ещё токенов)")
    print(f"  Остаток буфера  : {rest_3}  (буфер сохранён)")

    # ── Сценарий 4: реальный стрим — токены по одному ───────────────────────
    print(f"\n{'='*60}")
    print("[ Стрим ] Токены приходят по одному, force_every=5")
    print(f"{'='*60}")

    stream_buffer: List[int] = []
    output_chunks: List[str] = []

    for i, token_id in enumerate(all_ids):
        stream_buffer.append(token_id)
        print(f"  +токен[{i}]={token_id:>7}  буфер={stream_buffer}")

        chunk, stream_buffer = safe_stream_decode(
            adapter, stream_buffer, force_every=5
        )

        if chunk:
            output_chunks.append(chunk)
            print(f"  → выдали чанк   : {chunk!r}")
        else:
            print(f"  → копим (нет разделителя, буфер < 5)")

    # Хвост: принудительно декодируем всё, что осталось в буфере
    if stream_buffer:
        tail, _ = safe_stream_decode(adapter, stream_buffer, force_every=1)
        if tail:
            output_chunks.append(tail)
            print(f"\n  → хвост буфера  : {tail!r}")

    print(f"\nИтоговый текст: {''.join(output_chunks)!r}")
````
